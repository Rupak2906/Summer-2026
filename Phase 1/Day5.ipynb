**RECURSION**

```python
def multiply(a,b):
    result = 0
    for i in range(b):
        result = result + a

    print(result)
```

```python
multiply(3,2)
```

```python
def multiply(a,b):
    if b == 1:
        return a
    else:
        
        return a + multiply(a,b-1)
```

```python
multiply(3,12)
```

```python
def factorial(n):
    if n == 1:
        return 1

    else:
        return n*factorial(n-1)
```

```python
factorial(7)
```

**Palindrome Problem**

```python
def palindrome(str):
    palindrome_string = str[::-1]

    if str == palindrome_string:
        print("it is palindrome")
    else:
        print("Not palindrome")
```

```python
palindrome("madam")
```

```python
def palin(text):
    if len(text) <= 1:
        print("It's a palindrome")
    else:
        if text[0] == text[-1]:
            palindrome(text[1:-1])
        else:
            print("It's not a palindrome")
```

```python
palin("malayalam")
```

**Higher order function in Lambda**

```python
def return_sum(func,L):

    result = 0

    for i in L:
        if func(i):
            result = result + i
    return result

L = [12,53,78,31,43,73,44,22,68,45]

x = lambda x: x%2 == 0
y = lambda x: x%2 != 0
z = lambda x: x%3 == 0
        
```

```python
return_sum(x,L)
```

```python
return_sum(y,L)
```

```python
return_sum(z,L)
```
