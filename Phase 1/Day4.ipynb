**List**

- Lists are mutable unlike strings in python
- Lists are hetro and Arrays are homo
- Lists are slower but user friendly however arrays are faster but hard to arrange
- there can be 1D,2D,3D and so on lists and you can just access it by list[index]
- editing is easy just list[0] = 100
- adding is where it's fun, because list.append(1000) add at the end
- to add multiple things in the end of list use extend list.extend([2000,3000,4000])
- to add anywhere in the list use list.insert(index,what to add)
- delete has 4 ways - del, remove, pop, clear
- in del you gotta give the index to be remove
- in remove you gotta give the value to be removed
- in pop the last one gets removed and in clear the whole lists gets cleared

**Tuples**

- Compared to list to make a single element tuple we do it like this
       - tuple = ("Hello",) --- that is leave a comma after it
- Tuples are like strings -- immutable

**Sets**

- Sets don't allow duplicates
- Sets can't have mutable data types
- Sets don't have indexing/slicing
- Set itself is a mutable datatype
- Creating sets is by doing s1 = set()
- Set can be both homo and hetro
- Sets have no indexing as they index by the process of hashing

**Dictionary**

- Not mutable
- No indexing
- Keys are immutable, Values are mutable
- Keys should be unique
