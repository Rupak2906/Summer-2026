```python
class Atm:
    #Variables are always declared under init method
    #init is a constructor so whenever you make an object of atm class this gets executed
    #It's a magic method in python - these methods are not called by objects, they get called automatically
    #The double underscore before any object makes it private but not truly private

    #Creating Static/Class Variable - accessed by classname.variablename
    __counter = 1
    
    def __init__(self):
        #Instance Variable
        self.__pin = ""
        self.__balance = 0
        self.sno = Atm.__counter
        Atm.__counter = Atm.__counter + 1

    @staticmethod
    def get_counter(): #they don't have self in them and we use the staticmethod word for static methods
        return Atm.__counter

    @staticmethod
    def set_counter(new_counter):
        if type(new_counter) == int:
            Atm.__counter = new_counter
    def get_pin(self):
        return self.__pin

    def set_pin(self,new_pin):
        self.__pin = new_pin
        print("Pin changed")

    def menu(self):
        user_input = input("""
            Hi there! How would you like to proceed?
            1. Enter 1 to create a pin
            2. Enter 2 to deposit
            3. Enter 3 to withdraw money
            4. Enter 4 to check balance
            5. Enter 5 to exit
        """)

        if user_input == "1":
            self.create_pin()
        elif user_input == "2":
            self.deposit()
        elif user_input == "3":
            self.widthdraw()
        elif user_input == "4":
            self.check_balance()
        else:
            print("Bye")

    def create_pin(self):
        self.__pin = input("put in your pin")
        print("Successful")

    def deposit(self):
        temp = input("Enter your pin")
        if temp == self.__pin:
            deposit_money = int(input("Enter the amount to be deposited"))
            self.__balance = self.__balance + deposit_money
            print('deposited successfully')
        else:
            print("Invalid pin!")
    def widthdraw(self):
        temp = input("Enter your pin")
        if temp == self.__pin:
            deposit_money = int(input("Enter the amount to be widthdrawn"))
            if deposit_money <= self.__balance:
                self.__balance = self.__balance - deposit_money
                print("Widthdrawal Successful")
            else:
                print("Insufficient funds")
        else:
            print("Invalid pin!")

    def check_balance(self):
        temp = input("Enter your pin")
        if temp == self.__pin:
            print(self.__balance)
        else: 
            print("Incorrect Pin")
        
```

**Making my own class**

```python
class Fraction:

    def __init__(self,n,d):
        self.num  = n
        self.den = d

    def __str__(self):
        return "{}/{}".format(self.num,self.den)

    def __add__(self,other): #Here other is the second fraction given to add
        temp_num_1 = self.num * other.den
        temp_num_2 = self.den * other.num
        final_num = temp_num_1 + temp_num_2
        final_den = self.den * other.den

        return "{}/{}".format(final_num,final_den)

    def __sub__(self,other): #Here other is the second fraction given to add
        temp_num_1 = self.num * other.den
        temp_num_2 = self.den * other.num
        final_num = temp_num_1 - temp_num_2
        final_den = self.den * other.den

        return "{}/{}".format(final_num,final_den)

    def __mul__(self,other):
        final_num = self.num * other.num
        final_den = self.den * other.den

        return "{}/{}".format(final_num,final_den)

    def __truediv__(self,other):
        final_num = self.num * other.den
        final_den = self.den * other.num

        return "{}/{}".format(final_num,final_den)
```

```python
x = Fraction(3,5)
print(x)
```

```python
x = Fraction(3,0)
print(x)
```

```python
x = Fraction(3,2)
y = Fraction(4,5)
print(x+y)
```
