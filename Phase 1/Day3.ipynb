```python
import random
```

```python
correct = random.randint(1,100)

guess = int(input("guess a number between 1 to 100"))
attempts = 1

while guess != correct:
    if guess > correct:
        print("Guess lower")
    else:
        print("Guess higher")
    guess = int(input("guess a number between 1 to 100"))  
    attempts += 1

print("YAY, that's correct")
print("you took " , attempts,  "attempts")
        
```

```python
#Break, continue and Pass statements
```

```python
#Break - used for linear searching if finding a specific id in a loop, helps in decreasing run time
for i in range(1,11):
    if i == 9:
        break
    else:
        print(i)
```

```python
#Continue - instead of breaking the loop it just skips that specific condition
for i in range(1,11):
    if i == 5:
        continue
    else:
        print(i)
        print("Exists")
```

```python
#Pass - usually when you are unsure about what to write in the code you can put pass and that's read as come back later for python
for i in range(1,11):
    pass
```
